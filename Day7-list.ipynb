```python
#creating a list in python
# Empty list 
print([])

# 1-D list
print([1,2,3,4,5])

# 2-D list
print([1,2,3,[4,5,6]])

# 3-D list
print([1,2,3,['A','B',[4.5,9.8]]])
print([[[1,2],[4,5],[6,7]]])

# Hetrogenous List
print([1,True,5.5,'Mrinmoy'])

# using type Conversion
print(list('Mrinmoy'))
```

```python
# indexing in list
#positive indexing
L=[1,2,3,4,5]
print(L[0])
# nagative indexing
print(L[-2])

# 2d list indexing print 5
M= 1,2,3,[4,5]
print(M[-1][-1])
# indexing in 3-d lsit and print 2
N=[[[1,2],[3,4]],[[5,6],[6,7]]]
print(N[0][0][1])
```

```python
# slicing in list
L=[1,2,3,4,5,6,7]
print(L[0:5])
# jumps
print(L[0::2])
```

```python
# Adding item to a given list
# append
L=[1,2,3,4,5,6]
L.append(7) # use to add one item at a time  at the last position
print(L)

# extand
L.extend([7,8,9,'M'])# use to add multiple  item at the last position
print(L)

# insert
L.insert(1,100)# use to add item at our desiger location
print(L)
```
