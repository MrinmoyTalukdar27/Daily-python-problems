```python
# Simply importing the library 
import numpy as np
```

```python
# just generating random numbers for actual and predecited values
actual=np.random.randint(1,50,25)
predicated =np.random.randint(1,50,25)
```

```python
# here we are simplay implementing the mean squre error function
def mes(actual,predicated):
    return np.mean((actual-predicated)**2)
mes(actual,predicated)
```

#### Now we will see how we plot graphs in numpy

```python
import matplotlib.pyplot as plt
```

```python
# now we will take tow arrays of same size and plot a grapg of x=y
x=np.linspace(0,10,100)
y=x
```

```python
# now we will plota nd simple graph of two array 
plt.plot(x,y)
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('straight line graph')
plt.show()
```

```python
# now we will make a graph of x^2 (agraph of parabola)
x=np.linspace(-10,10,100)
y=x**2
plt.plot(x,y)
plt.xlabel('X-axis')
plt.ylabel('y-axis')
plt.title('parabola graph')
plt.show
```
